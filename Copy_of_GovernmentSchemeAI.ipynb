```python
def recommend_scheme(age, gender, occupation, income):

    recommendations = []

    occupation = occupation.lower()
    gender = gender.lower()

    for _, row in schemes.iterrows():

        scheme_occ = row["Occupation"].lower()
        scheme_gender = row["Gender"].lower()

        if age < row["MinAge"]:
            continue

        if scheme_gender != "any" and gender != scheme_gender:
            continue

        if scheme_occ != "any" and occupation != scheme_occ:
            continue

        if income > row["IncomeLimit"]:
            continue

        recommendations.append(row["Scheme"])

    return recommendations
```

```python
%%writefile app.py

import streamlit as st
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# -----------------------------
# Load Dataset
# -----------------------------
@st.cache_data
def load_data():
    return pd.read_csv(
        "/content/drive/MyDrive/GovSchemeAI/data/scheme_database.csv"
    )

scheme_db = load_data()

scheme_db["Search_Text"] = (
    scheme_db["Scheme Name"].fillna("") + " " +
    scheme_db["Description"].fillna("") + " " +
    scheme_db["Eligibility"].fillna("") + " " +
    scheme_db["Benefits"].fillna("")
)

vectorizer = TfidfVectorizer(stop_words="english")
tfidf_matrix = vectorizer.fit_transform(scheme_db["Search_Text"])

# -----------------------------
# Recommendation Function
# -----------------------------
def recommend(query):

    user_vector = vectorizer.transform([query])

    similarity = cosine_similarity(
        user_vector,
        tfidf_matrix
    )

    indices = similarity.argsort()[0][-5:][::-1]

    return scheme_db.iloc[indices]

# -----------------------------
# UI
# -----------------------------
st.title("🏛 Government Scheme Eligibility Assistant")

st.write(
    "Enter your details to find suitable Government Schemes."
)

query = st.text_area(
    "Describe yourself",
    placeholder="Example: Female engineering student from Karnataka with annual income 2 lakh"
)

if st.button("Find Schemes"):

    if query.strip() == "":
        st.warning("Please enter your details.")
        st.stop()

    results = recommend(query)

    st.success("Top Matching Schemes")

    for _, row in results.iterrows():

        st.subheader(row["Scheme Name"])

        st.markdown("**Eligibility**")
        st.write(row["Eligibility"])

        st.markdown("**Benefits**")
        st.write(row["Benefits"])

        st.markdown("**Required Documents**")
        st.write(row["Required Documents"])

        st.divider()
```

```python
def recommend(query, top_n=5, min_similarity=0.15, mention_disability=False):
    user_vector = vectorizer.transform([query])
    sims = cosine_similarity(user_vector, tfidf_matrix).flatten()
    df = scheme_db.assign(Score=sims)

    # Exclude disability-specific schemes unless the user actually mentioned it
    if not mention_disability:
        df = df[~df["Search_Text"].str.contains(r"disabilit", case=False, na=False)]

    df = df[df["Score"] >= min_similarity].sort_values("Score", ascending=False).head(top_n)
    return df
```

```python
%%writefile app.py

import streamlit as st
import pandas as pd
import joblib
import plotly.express as px
from sklearn.metrics.pairwise import cosine_similarity

st.set_page_config(page_title="Scheme Eligibility Assistant", page_icon="🏛", layout="wide")

st.markdown("""
<style>
.hero {
    background: linear-gradient(135deg, #2563eb, #7c3aed);
    padding: 40px 30px;
    border-radius: 16px;
    color: white;
    margin-bottom: 25px;
}
.hero h1 { font-size: 34px; margin-bottom: 6px; }
.hero p { font-size: 16px; opacity: 0.9; }
.scheme-card {
    background-color: #ffffff;
    border: 1px solid #e5e7eb;
    border-left: 6px solid #2563eb;
    border-radius: 12px;
    padding: 20px 24px;
    margin-bottom: 18px;
    box-shadow: 0 2px 6px rgba(0,0,0,0.04);
    transition: transform 0.15s ease;
}
.scheme-card:hover { transform: translateY(-2px); box-shadow: 0 6px 14px rgba(0,0,0,0.08); }
.scheme-title { font-size: 21px; font-weight: 700; color: #111827; margin-bottom: 4px; }
.section-label { font-weight: 600; color: #2563eb; margin-top: 10px; font-size: 13px; text-transform: uppercase; letter-spacing: 0.5px; }
.match-badge {
    display: inline-block;
    background: #dbeafe;
    color: #1e40af;
    padding: 3px 10px;
    border-radius: 20px;
    font-size: 12px;
    font-weight: 600;
    margin-left: 8px;
}
</style>
""", unsafe_allow_html=True)

@st.cache_resource
def load_models():
    vectorizer = joblib.load("/content/drive/MyDrive/GovSchemeAI/models/vectorizer.pkl")
    scheme_db = joblib.load("/content/drive/MyDrive/GovSchemeAI/models/scheme_db.pkl")
    tfidf_matrix = vectorizer.transform(scheme_db["Search_Text"])
    return vectorizer, scheme_db, tfidf_matrix

vectorizer, scheme_db, tfidf_matrix = load_models()

def recommend(query, top_n=5, min_similarity=0.15):
    user_vector = vectorizer.transform([query])
    sims = cosine_similarity(user_vector, tfidf_matrix).flatten()
    df = scheme_db.assign(Score=sims)
    df = df[df["Score"] >= min_similarity].sort_values("Score", ascending=False).head(top_n)
    return df

st.markdown("""
<div class="hero">
    <h1>🏛 Government Scheme Eligibility Assistant</h1>
    <p>Answer a few questions and discover Indian government welfare schemes you may qualify for.</p>
</div>
""", unsafe_allow_html=True)

st.sidebar.header("👤 Your Details")
age = st.sidebar.slider("Age", 0, 100, 25)
gender = st.sidebar.selectbox("Gender", ["Any", "Male", "Female", "Other"])
occupation = st.sidebar.text_input("Occupation", placeholder="e.g. Student, Farmer")
income = st.sidebar.number_input("Annual Income (₹)", min_value=0, value=200000, step=10000)
free_text = st.sidebar.text_area("Anything else? (optional)", placeholder="e.g. disabled veteran, rural area")
find_clicked = st.sidebar.button("🔍 Find My Schemes", use_container_width=True)

tab1, tab2, tab3 = st.tabs(["🔍 Search Results", "📋 Browse All Schemes", "ℹ️ About"])

with tab1:
    if find_clicked:
        query_parts = [occupation.strip(), gender if gender != "Any" else "", free_text.strip()]
        query = " ".join(p for p in query_parts if p)

        if not query:
            st.warning("Please enter at least your occupation or a short description of yourself.")
        else:
            with st.spinner("Matching you with schemes..."):
                results = recommend(query)

            if results.empty:
                st.warning("No confident matches found. Your dataset may not include schemes for this profile, or try rephrasing your details.")
            else:
                st.success(f"🎉 Found {len(results)} matching scheme(s) for you!")
                st.balloons()

                fig = px.bar(
                    results.sort_values("Score"),
                    x="Score", y="Scheme Name", orientation="h",
                    color="Score", color_continuous_scale="Blues",
                    labels={"Score": "Match Strength"},
                    height=280
                )
                fig.update_layout(margin=dict(l=0, r=0, t=10, b=0), coloraxis_showscale=False)
                st.plotly_chart(fig, use_container_width=True)

                for _, row in results.iterrows():
                    match_pct = int(row["Score"] * 100)
                    with st.container():
                        st.markdown(f"""
                        <div class="scheme-card">
                            <div class="scheme-title">{row['Scheme Name']}
                                <span class="match-badge">{match_pct}% match</span>
                            </div>
                        """, unsafe_allow_html=True)
                        st.progress(min(row["Score"] * 2, 1.0))
                        with st.expander("View Eligibility, Benefits & Documents"):
                            st.markdown(f"**Eligibility:** {row['Eligibility']}")
                            st.markdown(f"**Benefits:** {row['Benefits']}")
                            st.markdown(f"**Required Documents:** {row['Required Documents']}")
                        st.markdown("</div>", unsafe_allow_html=True)
    else:
        st.info("👈 Fill in your details in the sidebar and click **Find My Schemes** to get started.")

with tab2:
    st.subheader("All Available Schemes")
    search = st.text_input("Filter by keyword")
    display_df = scheme_db[["Scheme Name", "Eligibility", "Benefits"]]
    if search:
        display_df = display_df[display_df.apply(lambda r: search.lower() in str(r).lower(), axis=1)]
    st.dataframe(display_df, use_container_width=True, height=500)

with tab3:
    st.subheader("About this project")
    st.markdown("""
    This assistant uses **TF-IDF text similarity** combined with your personal details
    to recommend Indian government welfare schemes you may be eligible for.

    **Tech stack:** Python · Pandas · Scikit-learn · Streamlit · Plotly
    """)
    st.metric("Total Schemes in Database", len(scheme_db))
```
